2. Motor Lógico y Fundamentos (La base que mueve todo)
Sin estos tres pilares no se puede hacer Machine Learning de verdad:

Programación → principalmente Python (el lenguaje dominante).
Estadística → descriptiva + inferencial (media, varianza, pruebas de hipótesis, distribuciones, etc.).
Álgebra lineal → matrices, vectores, operaciones. Es lo que permite que “el robot se mueva” (modelos, transformaciones, optimización).

Conceptos importantes de sistemas (consistencia y disponibilidad) también aparecen cuando empiezas a trabajar con datos reales a escala.
Idea poderosa: las matemáticas + Python se reciclan constantemente. No son solo teoría: son las herramientas para limpiar, transformar y optimizar datos.
Herramientas de esta etapa:

Python + NumPy + Pandas
SQL (imprescindible)
Estadística básica + álgebra lineal aplicada
Git + GitHub


3. Capa de Datos: Estructura, Machine Learning y Minería de Datos
Los datos tienen estructura. Sobre esa estructura se construyen dos grandes áreas:

Minería de datos → descubrir patrones, relaciones y conocimiento oculto.
Machine Learning → enseñar a las máquinas a predecir o clasificar.

¿Para qué sirve en una empresa?

Predecir el futuro (ventas, churn, demanda…).
Clasificar (¿cliente feliz o triste?, ¿fraude o no?, ¿producto recomendado o no?).
Tomar decisiones basadas en evidencia en lugar de intuición.

Aquí es donde empiezas a construir modelos supervisados y no supervisados, a evaluarlos correctamente y a entender el ciclo completo: problema de negocio → datos → modelo → evaluación → impacto.

4. Infraestructura, Visualización y Big Data
De nada sirve un modelo brillante si:

Los datos están sucios.
Nadie entiende los resultados.
El sistema no escala.

Flujo real:
Datos sucios → limpieza y transformación → modelos → visualización y storytelling → decisión de negocio.
Cuando los datos crecen mucho aparece el Big Data: sistemas de archivos masivos distribuidos en cientos o miles de servidores (HDFS, cloud storage, etc.).
La línea de herramientas va desde lo simple hasta lo industrial:

Excel / Google Sheets → SQL → Python → herramientas de BI (Power BI, Tableau, Looker) → Spark / sistemas distribuidos → Java / Scala / cloud (según el rol).

Regla de oro: datos limpios + buena visualización = la empresa empieza a producir más valor. Datos sucios empeoran las decisiones.
